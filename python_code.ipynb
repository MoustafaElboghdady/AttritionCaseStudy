```python
import pandas as pd
from sqlalchemy import create_engine
```

```python
DB_CONFIG = {
    'host': 'localhost',
    'port': '5432',
    'database': 'attrition_case_study',
    'user': 'postgres',
    'password': 'Alahly1907'
}
 
def get_engine():
  
    url = (
        f"postgresql://{DB_CONFIG['user']}:{DB_CONFIG['password']}"
        f"@{DB_CONFIG['host']}:{DB_CONFIG['port']}/{DB_CONFIG['database']}"
    )
    return create_engine(url)
```

```python
df = pd.read_csv('F:\\WORK\\TP Case Study\\GBS BI HUB - BI Developer - HR Attrition Case Study (1).csv')
engine = get_engine()
df.sample(10)
```

```python
df.columns
```

```python
#df = df.copy()
df['ExperienceLevel'] = df['TotalWorkingYears'].apply(
    lambda y: 'Junior' if y < 5 else ('Mid' if y < 10 else 'Senior')
)

summary_table = df['ExperienceLevel'].value_counts().reset_index()
summary_table.columns = ['ExperienceLevel', 'EmployeeCount']

print(summary_table)
```

```python
df.to_sql("hr_attrition_data", con=engine, if_exists='append', index=False)
```
